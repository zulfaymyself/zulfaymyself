Question 10 — Score Checker

Create a function called check_score(score).

It should:

print "Passed" if score ≥ 50
print "Failed" otherwise

Test it with:

35, 50, 78, 42, 91

```python
scores = [35,50,78,42,91]

def check_score(scores):
    for score in scores:
        if score >= 50:
            print(score, "passed")
        else:
            print(score, "failed")
check_score(scores)
```

Question 13 — Analyze Student Marks

Given: marks = [45, 78, 62, 91, 38, 55, 84, 73, 49, 88]

Write a program that determines:

Total marks
Average marks
Highest mark
Lowest mark
Number of students who passed
Number of students who failed

Use a for loop to count the passed and failed students.

```python
marks = [45, 78, 62, 91, 38, 55, 84, 73, 49, 88]
average = sum(marks) / len(marks)

passed = []
failed = []

for mark in marks:
        if mark >= 50:
            passed.append(mark)
        else:
            failed.append(mark)

print(sum(marks))
print(average)
print(max(marks))
print(min(marks))
print(len(passed))
print(len(failed))
```
